# Tutorial de Uso do Google Colab

Este tutorial irá guiá-lo pelos conceitos básicos de como usar o Google Colab para criar e executar arquivos Python.

## 1. O que é Google Colab?

O Google Colaboratory (Colab) é um ambiente de notebook Jupyter hospedado no Google Cloud. Ele permite que você escreva e execute código Python diretamente no seu navegador, sem necessidade de configuração, e oferece acesso gratuito a GPUs e TPUs. É amplamente utilizado para aprendizado de máquina, análise de dados e educação.

## 2. Criando um Novo Notebook

Para começar, você precisará criar um novo notebook Colab:

1.  Vá para [colab.research.google.com](https://colab.research.google.com/).
2.  Clique em "Arquivo" > "Novo notebook".

Isso abrirá um novo notebook vazio onde você pode começar a escrever seu código.

## 3. Células de Código e Texto

Notebooks Colab são compostos por células. Existem dois tipos principais:

*   **Células de Código (Code cells)**: Para escrever e executar código Python.
*   **Células de Texto (Text cells)**: Para escrever texto formatado usando Markdown, como este tutorial.

Você pode adicionar novas células clicando nos botões `+ Código` ou `+ Texto` na barra de ferramentas superior ou passando o mouse entre as células existentes.

## 4. Escrevendo e Executando Código Python

### 4.1. Escrevendo Código

Em uma célula de código, você pode digitar qualquer código Python. Por exemplo:

```python
print("Olá, Colab!")
x = 10
y = 200
z = 150
print(f"A soma de x e y é: {x + y}")
```

```python
print(z)
```

### 4.2. Executando Código

Para executar uma célula de código, você tem algumas opções:

*   Clique no ícone de "Play" (triângulo) à esquerda da célula.
*   Use o atalho de teclado `Shift + Enter` (executa a célula e seleciona a próxima).
*   Use o atalho de teclado `Ctrl + Enter` (executa a célula e mantém a célula atual selecionada).

Os resultados do seu código (saída padrão, erros, etc.) aparecerão logo abaixo da célula.

## 5. Salvando seu Notebook

Seu notebook é salvo automaticamente no seu Google Drive. Você também pode salvar manualmente:

*   **Arquivo** > **Salvar** ou `Ctrl + S`.
*   **Arquivo** > **Salvar uma cópia no Drive** (para criar uma cópia).
*   **Arquivo** > **Fazer o download** (para baixar o notebook como um arquivo `.ipynb` ou `.py`).

## 6. Criando e Gerenciando Arquivos Python (`.py`)

Às vezes, você pode querer criar arquivos Python separados (`.py`) dentro do ambiente Colab, seja para organizar seu código ou para simular um ambiente de desenvolvimento local.

### 6.1. Criando um arquivo `.py`

Você pode criar um novo arquivo `.py` usando o comando `%%writefile` em uma célula de código:

```python
%%writefile meu_modulo.py
def saudacao(nome):
    return f"Olá, {nome}! Bem-vindo ao Colab."

if __name__ == '__main__':
    print(saudacao("Mundo"))
```

Ao executar esta célula, um arquivo chamado `meu_modulo.py` será criado no diretório de conteúdo da sua sessão Colab.

### 6.2. Visualizando e Gerenciando Arquivos

Você pode acessar o explorador de arquivos clicando no ícone de pasta na barra lateral esquerda. Lá você pode:

*   **Ver seus arquivos**: `meu_modulo.py` estará lá.
*   **Fazer upload**: Clique no ícone de upload para carregar arquivos do seu computador.
*   **Baixar**: Clique nos três pontos ao lado do nome do arquivo e selecione "Fazer download".
*   **Excluir/Renomear**: Use as opções disponíveis ao clicar nos três pontos.

## 7. Executando Arquivos Python (`.py`)

Existem algumas maneiras de executar um arquivo `.py` que você criou ou carregou no Colab:

### 7.1. Usando `%run` (Magic Command)

O comando mágico `%run` executa um script Python como se ele tivesse sido digitado na linha de comando.

```python
%run meu_modulo.py
```

Isso executará o código dentro de `meu_modulo.py`. Se `meu_modulo.py` contiver a linha `if __name__ == '__main__':`, esse bloco será executado.

### 7.2. Importando como Módulo

Você também pode importar o arquivo `.py` como um módulo Python regular. Isso é útil se você tem funções ou classes que deseja reutilizar em seu notebook.

```python
import meu_modulo

print(meu_modulo.saudacao("Usuário Colab"))
```

### 7.3. Executando com `bash` (para scripts mais complexos ou argumentos)

Para cenários mais avançados, onde você precisa passar argumentos de linha de comando ou simular um ambiente de terminal, você pode usar o bash.

Para utilizar comandos bash utilize ! no começo do comando.

## Teste de Nivelamento


```python
%%writefile teste_nivelamento.py

import random
import math
import unittest
import pandas as pd

def gerar_datasets():
    """Gera duas listas de dicionários simulando tabelas de um banco de dados."""
    # A seed fixa ajuda na reprodutibilidade, mas os testes abaixo não dependem
    # de valores fixados no código (hardcoded), usando o Pandas para validação.
    random.seed(42)
    n = 200
    ids = list(range(1, n + 1))

    tabela_notas = []
    for i in ids:
        tabela_notas.append({
            'aluno_id': i,
            'horas_estudo': round(random.gauss(10, 3), 1),
            'nota_programacao': round(random.gauss(7, 1.5), 1),
            'turno': random.choice(['Matutino', 'Noturno'])
        })

    # Inserindo valores nulos (None) propositais
    for i in range(10, 20):
        tabela_notas[i]['nota_programacao'] = None

    ids_embaralhados = ids[:]
    random.shuffle(ids_embaralhados)
    tabela_presenca = []
    for i in ids_embaralhados:
        tabela_presenca.append({
            'aluno_id': i,
            'presenca': round(random.gauss(80, 10), 1)
        })

    return tabela_notas, tabela_presenca

# =======================================================
# ÁREA DO ALUNO - PREENCHA AS FUNÇÕES ABAIXO EM PYTHON PURO
# =======================================================

def questao_1_manipular_dados(notas, presencas):
    """
    Realize um INNER JOIN manual entre as duas listas de dicionários
    usando a chave 'aluno_id'.
    Filtre o resultado mantendo APENAS alunos com 'presenca' >= 75.0.
    Retorne uma nova lista de dicionários com os dados combinados.
    """
    resultado = []

    for nota in notas:
        for presenca in presencas:
            if nota["aluno_id"] == presenca["aluno_id"]:
                if presenca["presenca"] >= 75.0:
                    aluno = nota.copy()
                    aluno.update(presenca)
                    resultado.append(aluno)

    return resultado

def questao_2_estatistica(tb_notas):
    """
    Calcule a média e o desvio padrão amostral da chave 'horas_estudo'.
    Dica: o desvio padrão amostral divide pela quantidade (N - 1).
    Retorne a tupla: (media, desvio_padrao).
    """
    horas = []

    for aluno in tb_notas:
      horas.append(aluno["horas_estudo"])

    media = sum(horas) / len(horas)
    soma = 0

    for hora in horas:
      soma += (hora - media) ** 2

    desvio_padrao = math.sqrt(soma / (len(horas) - 1))

    return media, desvio_padrao

def questao_3_divisao_manual(dados, proporcao_teste=0.2):
    """
    Divida a lista de dados em treino e teste MANUALMENTE usando fatiamento (slicing).
    Separe os primeiros elementos para treino e o final para teste.
    Retorne a tupla: (lista_treino, lista_teste).
    """
    tamanho_teste = int(len(dados) * proporcao_teste)
    ponto_divisao = len(dados) - tamanho_teste
    lista_treino = dados [:ponto_divisao]
    lista_teste = dados[ponto_divisao:]

    return lista_treino, lista_teste

def questao_4_dados_nulos(tb_notas):
    """
    Identifique os valores nulos (None) na chave 'nota_programacao'
    e substitua-os pela média das notas válidas desta mesma chave.
    Retorne a lista atualizada.
    """
    notas_validas = []

    for aluno in tb_notas:
      if aluno["nota_programacao"] is not None:
        notas_validas.append(aluno["nota_programacao"])

    media = sum(notas_validas) / len(notas_validas)

    for aluno in tb_notas:
      if aluno ["nota_programacao"] is None:
        aluno["nota_programacao"] = media

    return tb_notas

def questao_5_calculo_erro(y_true, y_pred):
    """
    Recebe duas listas de mesmo tamanho: valores reais e predições.
    Calcule o Erro Quadrático Médio (MSE).
    Retorne o valor numérico do MSE.
    """
    soma_erros = 0

    for real, previsao in zip(y_true, y_pred):
      error = real - previsao
      soma_erros += error ** 2

    mse = soma_erros /len(y_true)

    return mse

def questao_6_agrupamento(tb_notas):
    """
    Agrupe os dados pela chave 'turno' e calcule a média de
    'horas_estudo' para cada turno.
    Retorne um dicionário no formato: {'Matutino': media, 'Noturno': media}.
    """
    grupos = {}

    for aluno in tb_notas:
      turno = aluno["turno"]
      horas = aluno["horas_estudo"]

      if turno not in grupos:
        grupos[turno] = {
            "soma": 0,
            "quantidade": 0
        }
      grupos[turno]["soma"] += horas
      grupos[turno]["quantidade"] += 1

    resultado = {}

    for turno in grupos:
      resultado[turno] = grupos[turno]["soma"] / grupos[turno]["quantidade"]

    return resultado

# =======================================================
# TESTES DE VALIDAÇÃO AUTOMÁTICA (NÃO ALTERE ESTA PARTE)
# =======================================================
class TestNivelamentoPythonPuro(unittest.TestCase):
    def setUp(self):
        # Gerando cópias novas para cada teste
        self.tb_notas, self.tb_presenca = gerar_datasets()
        # Convertendo para DataFrames estritamente para o cálculo do gabarito dinâmico
        self.df_notas = pd.DataFrame(self.tb_notas)
        self.df_presenca = pd.DataFrame(self.tb_presenca)

    def test_q1_manipulacao(self):
        try:
            # Gabarito Dinâmico
            df_merged = pd.merge(self.df_notas, self.df_presenca, on='aluno_id', how='inner')
            df_esperado = df_merged[df_merged['presenca'] >= 75.0]

            resultado = questao_1_manipular_dados(self.tb_notas, self.tb_presenca)

            self.assertIsNotNone(resultado)
            self.assertTrue(isinstance(resultado, list))
            self.assertEqual(len(resultado), len(df_esperado), "A quantidade de alunos após o join e filtro está incorreta.")
            self.assertIn('presenca', resultado[0], "A chave 'presenca' não foi encontrada após o Join.")
            self.assertTrue(all(d.get('presenca', 0) >= 75.0 for d in resultado), "Há alunos com presença inferior a 75.0.")
        except Exception as e: self.fail(f"Erro na Questão 1: {e}")

    def test_q2_estatistica(self):
        try:
            # Gabarito Dinâmico
            esperado_media = self.df_notas['horas_estudo'].mean()
            esperado_std = self.df_notas['horas_estudo'].std(ddof=1)

            media, std = questao_2_estatistica(self.tb_notas)

            self.assertAlmostEqual(media, esperado_media, places=2, msg="A média está incorreta.")
            self.assertAlmostEqual(std, esperado_std, places=2, msg="O desvio padrão está incorreto.")
        except Exception as e: self.fail(f"Erro na Questão 2: {e}")

    def test_q3_divisao(self):
        try:
            treino, teste = questao_3_divisao_manual(self.tb_notas, 0.2)

            tamanho_esperado_teste = int(len(self.tb_notas) * 0.2)
            self.assertEqual(len(treino) + len(teste), len(self.tb_notas), "A soma do treino e teste não bate com o total.")
            self.assertEqual(len(teste), tamanho_esperado_teste, "A proporção do conjunto de teste está incorreta.")
        except Exception as e: self.fail(f"Erro na Questão 3: {e}")

    def test_q4_nulos(self):
        try:
            # Gabarito Dinâmico (o Pandas ignora NaNs no cálculo da média automaticamente)
            media_esperada = self.df_notas['nota_programacao'].mean()

            resultado = questao_4_dados_nulos(self.tb_notas)
            nulos = [d for d in resultado if d['nota_programacao'] is None]

            self.assertEqual(len(nulos), 0, "Ainda existem valores nulos (None) na lista.")
            # O índice 15 era sabidamente nulo na geração dos dados
            self.assertAlmostEqual(resultado[15]['nota_programacao'], media_esperada, places=2,
                                   msg="A substituição dos valores nulos não utilizou a média correta.")
        except Exception as e: self.fail(f"Erro na Questão 4: {e}")

    def test_q5_calculo(self):
        try:
            # Gerando listas randômicas para testar
            y_true = [random.uniform(0, 10) for _ in range(15)]
            y_pred = [random.uniform(0, 10) for _ in range(15)]

            # Gabarito Dinâmico
            serie_true = pd.Series(y_true)
            serie_pred = pd.Series(y_pred)
            mse_esperado = ((serie_true - serie_pred) ** 2).mean()

            res = questao_5_calculo_erro(y_true, y_pred)
            self.assertAlmostEqual(res, mse_esperado, places=3, msg="O cálculo do MSE falhou.")
        except Exception as e: self.fail(f"Erro na Questão 5: {e}")

    def test_q6_agrupamento(self):
        try:
            # Gabarito Dinâmico
            agrupado_esperado = self.df_notas.groupby('turno')['horas_estudo'].mean().to_dict()

            res = questao_6_agrupamento(self.tb_notas)

            self.assertTrue(isinstance(res, dict), "O retorno deve ser um dicionário.")
            self.assertIn('Matutino', res)
            self.assertIn('Noturno', res)
            self.assertAlmostEqual(res['Matutino'], agrupado_esperado['Matutino'], places=2)
            self.assertAlmostEqual(res['Noturno'], agrupado_esperado['Noturno'], places=2)
        except Exception as e: self.fail(f"Erro na Questão 6: {e}")

if __name__ == '__main__':
    print("\n=== TESTE DE NIVELAMENTO - AVALIAÇÃO DINÂMICA ===")
    unittest.main(verbosity=2)
```
